# Test PySpark - Vérification de l'Installation

Ce notebook vérifie que PySpark est bien installé et fonctionnel.

## 1. Initialiser Spark Session

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

# Créer une Spark Session
spark = SparkSession.builder \
    .appName("Test PySpark") \
    .master("local[*]") \
    .getOrCreate()

print("✅ Spark Session créée!")
print(f"Version de Spark : {spark.version}")
```

## 2. Créer un DataFrame de Test

```python
# Créer un DataFrame simple
data = [("Alice", 25), ("Bob", 30), ("Charlie", 35)]
columns = ["name", "age"]

df = spark.createDataFrame(data, columns)

print("✅ DataFrame créé!")
df.show()
```

## 3. Test de Lecture CSV

```python
# Tester la lecture d'un fichier CSV
df_test = spark.read.csv('../CSVs/data_merged.csv', header=True, inferSchema=True, sep=';')

print("✅ Fichier CSV chargé!")
print(f"Nombre de lignes : {df_test.count()}")
print(f"Nombre de colonnes : {len(df_test.columns)}")
print("\nColonnes :")
df_test.printSchema()
print("\nAperçu :")
df_test.show(5)
```

## 4. Test d'Opérations Spark

```python
# Test de quelques opérations
df_filtered = df_test.filter(col("nb_velos") > 5000)

print(f"✅ Filtrage réussi!")
print(f"Lignes avec > 5000 vélos : {df_filtered.count()}")
df_filtered.show(5)
```

## 5. Arrêter Spark

```python
# Toujours arrêter la session Spark à la fin
spark.stop()
print("✅ Spark Session arrêtée!")
```

## ✅ Si toutes les cellules s'exécutent sans erreur, PySpark est correctement installé !
